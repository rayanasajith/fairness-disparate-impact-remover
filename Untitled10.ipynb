```python
import os, urllib.request, warnings
import numpy as np
import pandas as pd
import aif360
from aif360.datasets import AdultDataset, GermanDataset, CompasDataset
warnings.filterwarnings("ignore")
print("AIF360 version:", aif360.__version__)
```

```python


RAW_FILES = {
    # UCI Adult: original train file, test file and description
    "adult": ["https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.data",
              "https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.test",
              "https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.names"],
    # UCI Statlog German Credit: data file and description
    "german": ["https://archive.ics.uci.edu/ml/machine-learning-databases/statlog/german/german.data",
               "https://archive.ics.uci.edu/ml/machine-learning-databases/statlog/german/german.doc"],
    # ProPublica COMPAS two-year recidivism data
    "compas": ["https://raw.githubusercontent.com/propublica/compas-analysis/master/compas-scores-two-years.csv"],
}

# Location of AIF360's raw-data folder inside the installed package
raw_root = os.path.join(os.path.dirname(aif360.__file__), "data", "raw")

for name, urls in RAW_FILES.items():
    folder = os.path.join(raw_root, name)
    os.makedirs(folder, exist_ok=True)              # create the folder if needed
    for url in urls:
        dest = os.path.join(folder, os.path.basename(url))
        if not os.path.exists(dest):                # skip files we already have
            print("Downloading", os.path.basename(url))
            urllib.request.urlretrieve(url, dest)

print("Raw data ready in", raw_root)
```

```python

DATASETS = {
    # Adult: predict income > 50K; protected = sex (Male privileged).
    # Same five numeric features as the AIF360 demo.
    "Adult": dict(cls=AdultDataset, protected="sex", privileged=[["Male"]],
                  features=["age", "education-num", "capital-gain",
                            "capital-loss", "hours-per-week"]),
    # German Credit: predict good credit; protected = age, split at 25
    # (as in the paper, following Kamiran & Calders). Age >= 25 is privileged.
    "German": dict(cls=GermanDataset, protected="age", privileged=[lambda x: x >= 25],
                   features=["month", "credit_amount", "investment_as_income_percentage",
                             "residence_since", "number_of_credits", "people_liable_for"]),
    # COMPAS: favourable outcome = no reoffending within two years;
    # protected = race (Caucasian privileged, all other races unprivileged).
    "COMPAS": dict(cls=CompasDataset, protected="race", privileged=[["Caucasian"]],
                   features=["age", "juv_fel_count", "juv_misd_count",
                             "juv_other_count", "priors_count"]),
}


def load(name):
    """Load one dataset as an AIF360 BinaryLabelDataset.

    Returns the dataset plus the privileged / unprivileged group definitions
    in the format AIF360's metric classes expect.
    """
    cfg = DATASETS[name]
    ds = cfg["cls"](protected_attribute_names=[cfg["protected"]],
                    privileged_classes=cfg["privileged"],
                    categorical_features=[],              # no one-hot encoding: numeric only
                    features_to_keep=cfg["features"])     # protected attr + label are kept automatically
    # After loading, AIF360 encodes the protected attribute as 1 = privileged, 0 = unprivileged
    priv = [{cfg["protected"]: 1.0}]
    unpriv = [{cfg["protected"]: 0.0}]
    return ds, priv, unpriv


# Quick overview of each dataset
for name in DATASETS:
    ds, priv, unpriv = load(name)
    p = ds.protected_attributes.ravel()                   # 1 = privileged, 0 = unprivileged
    y = (ds.labels.ravel() == ds.favorable_label)         # True = favourable outcome
    print(f"\n{name}: n = {len(y)}")
    print("  features:", ds.feature_names)
    print(f"  favourable rate = {y.mean():.3f} | privileged share = {p.mean():.3f}")
```

```python
from aif360.metrics import BinaryLabelDatasetMetric
from sklearn.neighbors import NearestNeighbors
from sklearn.preprocessing import StandardScaler

K_NEIGHBORS = 5   # number of neighbours for the consistency measure (AIF360's default)


def consistency(ds, labels, k=K_NEIGHBORS):
    """Individual-fairness measure from Zemel et al. (2013).

    consistency = 1 - (1/n) * sum_i | y_i - mean label of the k nearest neighbours of x_i |

    1 means every person gets the same outcome as the people most similar to them.
    Design choices (state these in the report):
      - neighbours are found on STANDARDISED features, so large-valued features
        such as capital-gain do not dominate the distance;
      - the protected attribute is EXCLUDED, so "similar" means similar
        in everything except group membership;
      - each point is excluded from its own neighbourhood.
    """
    # Indices of all features except the protected attribute
    keep = [i for i, f in enumerate(ds.feature_names)
            if f not in ds.protected_attribute_names]
    X = StandardScaler().fit_transform(ds.features[:, keep])

    # Binary outcome vector: 1 = favourable, 0 = unfavourable
    y = (np.asarray(labels).ravel() == ds.favorable_label).astype(float)

    # Ask for k+1 neighbours because the closest one is usually the point itself
    _, idx = NearestNeighbors(n_neighbors=k + 1).fit(X).kneighbors(X)

    # Remove the point itself from its neighbour list, keep the first k others
    not_self = idx != np.arange(len(y))[:, None]
    neigh = np.array([row[m][:k] for row, m in zip(idx, not_self)])

    return 1 - np.mean(np.abs(y - y[neigh].mean(axis=1)))


def fairness_metrics(ds, labels, priv, unpriv):
    """Group and individual fairness of a set of labels.

    `labels` can be the true labels (before mitigation) or a classifier's
    predictions (after mitigation). DP_Diff and DIR depend only on the labels
    and group membership; consistency also uses the features of `ds`.
    """
    # Work on a copy so the original dataset is never modified
    d = ds.copy(deepcopy=True)
    d.labels = np.asarray(labels, dtype=float).reshape(-1, 1)
    d.instance_weights = np.ones(len(d.labels))   # unweighted evaluation

    m = BinaryLabelDatasetMetric(d, unprivileged_groups=unpriv, privileged_groups=priv)
    return {
        # P(favourable | privileged) and P(favourable | unprivileged)
        "Rate_priv": m.base_rate(privileged=True),
        "Rate_unpriv": m.base_rate(privileged=False),
        # Demographic parity difference: Rate_unpriv - Rate_priv (0 = parity)
        "DP_Diff": m.statistical_parity_difference(),
        # Disparate impact ratio: Rate_unpriv / Rate_priv (1 = parity, < 0.8 fails the 80% rule)
        "DIR": m.disparate_impact(),
        # Individual fairness (1 = fully consistent)
        "Consistency": consistency(d, d.labels),
    }
```

```python
# Fairness of the ORIGINAL labels on each FULL dataset:
# no split, no classifier, no mitigation yet (assignment step 2).
before_rows = []
for name in DATASETS:
    ds, priv, unpriv = load(name)
    r = fairness_metrics(ds, ds.labels, priv, unpriv)   # true labels
    r["Dataset"] = name
    before_rows.append(r)

before_df = pd.DataFrame(before_rows).set_index("Dataset")
before_df.round(4)
```

```python
from sklearn.model_selection import train_test_split

TEST_SIZE = 0.3              # 70% train / 30% test
SEEDS = [0, 1, 2, 3, 4]      # each seed = a different random split (


def stratified_split(ds, seed, test_size=TEST_SIZE):
    """Stratified train/test split of an AIF360 dataset.

    We stratify on the COMBINATION of label and protected group (4 cells:
    privileged/unprivileged x favourable/unfavourable), so that both the
    base rates and the group-specific base rates are the same in train
    and test. Stratifying on the label alone could leave the small
    unprivileged group (e.g. ~150 young people in German) unevenly split.
    """
    p = ds.protected_attributes.ravel().astype(int)                 # 1 = privileged, 0 = unprivileged
    y = (ds.labels.ravel() == ds.favorable_label).astype(int)       # 1 = favourable, 0 = unfavourable
    strata = 2 * y + p                                              # cell id 0..3 for each row

    idx = np.arange(len(y))
    train_idx, test_idx = train_test_split(idx, test_size=test_size,
                                           stratify=strata, random_state=seed)
    # subset() returns new AIF360 datasets with the selected rows
    return ds.subset(train_idx), ds.subset(test_idx)
```

```python
# Sanity check for one seed: the favourable rates (overall and per group)
# and the privileged share should be (almost) identical in full / train / test.
check = []
for name in DATASETS:
    ds, _, _ = load(name)
    train, test = stratified_split(ds, SEEDS[0])
    for part, d in [("full", ds), ("train", train), ("test", test)]:
        p = d.protected_attributes.ravel()
        y = d.labels.ravel() == d.favorable_label
        check.append({"Dataset": name, "Part": part, "n": len(y),
                      "Fav. rate": y.mean(),
                      "Privileged share": p.mean(),
                      "Fav. rate priv": y[p == 1].mean(),
                      "Fav. rate unpriv": y[p == 0].mean()})

pd.DataFrame(check).set_index(["Dataset", "Part"]).round(4)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import MinMaxScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import accuracy_score
from aif360.algorithms.preprocessing import DisparateImpactRemover


REPAIR_LEVELS = np.round(np.linspace(0, 1, 11), 1)


def drop_protected(ds):
    """Feature matrix WITHOUT the protected attribute.
    The repair uses the protected attribute, but the classifier must not see it."""
    keep = [i for i, f in enumerate(ds.feature_names)
            if f not in ds.protected_attribute_names]
    return ds.features[:, keep]


def make_classifier(seed):
    """Min-max scaling (fitted on the training data only, as part of the pipeline)
    + logistic regression with balanced class weights, as in the paper and the AIF360 demo."""
    return make_pipeline(
        MinMaxScaler(),
        LogisticRegression(class_weight="balanced", solver="liblinear", random_state=seed))


def repair_and_predict(train, test, protected, level, seed):
    """Repair train and test at repair level `level`, train the classifier on the
    repaired training data and return its predictions for the repaired test data."""
    di = DisparateImpactRemover(repair_level=level, sensitive_attribute=protected)

    # AIF360 only offers fit_transform: each set is repaired using its own
    # group-wise feature distributions (same as the AIF360 demo).
    train_rep = di.fit_transform(train)
    test_rep = di.fit_transform(test)

    clf = make_classifier(seed)
    clf.fit(drop_protected(train_rep), train_rep.labels.ravel())
    y_pred = clf.predict(drop_protected(test_rep))
    return y_pred, train_rep, test_rep
```

```python

name = "Adult"
cfg = DATASETS[name]
ds, priv, unpriv = load(name)
train, test = stratified_split(ds, SEEDS[0])

for level in [0.0, 1.0]:
    y_pred, train_rep, test_rep = repair_and_predict(train, test, cfg["protected"], level, SEEDS[0])
    # Fairness is computed from the predictions; consistency uses the ORIGINAL
    # test features, so "similar people" means the same thing before and after repair.
    m = fairness_metrics(test, y_pred, priv, unpriv)
    acc = accuracy_score(test.labels.ravel(), y_pred)
    print(f"λ = {level}: accuracy = {acc:.4f} | DP_Diff = {m['DP_Diff']:.4f} | "
          f"DIR = {m['DIR']:.4f} | consistency = {m['Consistency']:.4f}")

# Sanity checks on the repair itself
rep0 = DisparateImpactRemover(repair_level=0.0, sensitive_attribute=cfg["protected"]).fit_transform(train)
rep1 = DisparateImpactRemover(repair_level=1.0, sensitive_attribute=cfg["protected"]).fit_transform(train)
print("\nλ = 0 leaves the features unchanged:", np.allclose(rep0.features, train.features))
print("Protected attribute untouched by the repair:",
      np.array_equal(rep1.protected_attributes, train.protected_attributes))
print("Labels untouched by the repair:", np.array_equal(rep1.labels, train.labels))

# Per-group feature means before and after full repair:
# after repair, the two groups should have (nearly) the same distribution per feature.
p = train.protected_attributes.ravel() == 1
rows = []
for j, f in enumerate(train.feature_names):
    if f in train.protected_attribute_names:
        continue
    rows.append({"feature": f,
                 "priv (before)": train.features[p, j].mean(),
                 "unpriv (before)": train.features[~p, j].mean(),
                 "priv (after)": rep1.features[p, j].mean(),
                 "unpriv (after)": rep1.features[~p, j].mean()})
pd.DataFrame(rows).set_index("feature").round(3)
```

```python
# Reproduce the example of Figure 1 in Feldman et al. (2015):
# SAT scores ~ N(400, 50) for males (privileged) and N(550, 100) for females.
# According to the paper, a full repair maps BOTH groups onto the "median"
# distribution, which for two groups is N(475, 75).
from aif360.datasets import BinaryLabelDataset

rng = np.random.default_rng(0)
n = 5000
sex = rng.integers(0, 2, n)                                    # 1 = male, 0 = female
sat = np.where(sex == 1, rng.normal(400, 50, n), rng.normal(550, 100, n))
y = rng.integers(0, 2, n)                                      # labels are irrelevant for the repair

toy = BinaryLabelDataset(df=pd.DataFrame({"sat": sat, "sex": sex, "y": y}),
                         label_names=["y"], protected_attribute_names=["sex"],
                         favorable_label=1, unfavorable_label=0)
toy_rep = DisparateImpactRemover(repair_level=1.0, sensitive_attribute="sex").fit_transform(toy)

j = toy.feature_names.index("sat")
g = toy.protected_attributes.ravel()
for val, label in [(1, "male (priv)"), (0, "female (unpriv)")]:
    m = g == val
    print(f"{label:16s} before: mean {toy.features[m, j].mean():6.1f}, sd {toy.features[m, j].std():5.1f}"
          f"  | after: mean {toy_rep.features[m, j].mean():6.1f}, sd {toy_rep.features[m, j].std():5.1f}")
print("Paper's expected result after full repair: both groups mean ≈ 475, sd ≈ 75")
```

```python

# If the repair works as in the paper, the 'after' rows of the two groups match.
qs = [0.1, 0.25, 0.5, 0.75, 0.9]
rows = []
for jf, f in enumerate(train.feature_names):
    if f in train.protected_attribute_names:
        continue
    for stage, data in [("before", train), ("after", rep1)]:
        for grp, mask in [("priv", p), ("unpriv", ~p)]:
            r = {"feature": f, "stage": stage, "group": grp,
                 "unique values": len(np.unique(data.features[mask, jf]))}
            r.update({f"q{int(q*100)}": np.quantile(data.features[mask, jf], q) for q in qs})
            rows.append(r)

# the repaired feature matrix must still match the original protected attribute.
k = train.feature_names.index(cfg["protected"])
print("Row order preserved:", np.array_equal(rep1.features[:, k], train.features[:, k]))

pd.DataFrame(rows).set_index(["feature", "stage", "group"]).round(2)
```

```python
from sklearn.metrics import balanced_accuracy_score
from tqdm import tqdm

rows = []
for name, cfg in DATASETS.items():
    ds, priv, unpriv = load(name)
    for seed in tqdm(SEEDS, desc=name):
        train, test = stratified_split(ds, seed)
        y_true = test.labels.ravel()

        # Reference: fairness of the TRUE labels of this test set
        r = fairness_metrics(test, y_true, priv, unpriv)
        r.update(Dataset=name, Seed=seed, Setting="Test labels", Level=np.nan)
        rows.append(r)

        # Predictions after repair at every level (λ = 0 is the unmitigated baseline)
        for level in REPAIR_LEVELS:
            y_pred, _, _ = repair_and_predict(train, test, cfg["protected"], level, seed)
            r = fairness_metrics(test, y_pred, priv, unpriv)   # consistency on ORIGINAL test features
            r.update(Accuracy=accuracy_score(y_true, y_pred),
                     Balanced_Acc=balanced_accuracy_score(y_true, y_pred))
            r.update(Dataset=name, Seed=seed, Setting="Predictions", Level=level)
            rows.append(r)

results = pd.DataFrame(rows)
results.to_csv("results_per_seed.csv", index=False)    # all raw numbers, for the repo
print("Done:", len(results), "rows")
```

```python
cols = ["DP_Diff", "DIR", "Consistency", "Accuracy", "Balanced_Acc"]
pred = results[results.Setting == "Predictions"]

# Baseline (λ = 0, no repair) and full repair (λ = 1)
sel = pred[pred.Level.isin([0.0, 1.0])].copy()
sel["Setting"] = np.where(sel.Level == 0.0, "2. Baseline LR (λ = 0)", "3. DI Remover + LR (λ = 1)")
agg = sel.groupby(["Dataset", "Setting"])[cols].agg(["mean", "std"])

# Format as "mean ± std"
fmt = pd.DataFrame({c: agg[(c, "mean")].map("{:.3f}".format) + " ± " + agg[(c, "std")].map("{:.3f}".format)
                    for c in cols})

# Add the "before" row: fairness of the original labels on the full dataset (Step 2)
before = before_df[["DP_Diff", "DIR", "Consistency"]].round(3).astype(str)
before["Setting"] = "1. Original labels (full data)"
before = before.reset_index().set_index(["Dataset", "Setting"])

summary = pd.concat([before, fmt]).sort_index().fillna("–")
summary.to_csv("results_summary.csv")
summary
```

```python
%matplotlib inline
import matplotlib.pyplot as plt

# Mean and std over seeds for every repair level
curve = pred.groupby(["Dataset", "Level"])[["DIR", "Accuracy", "Balanced_Acc"]].agg(["mean", "std"])

fig, axes = plt.subplots(2, 3, figsize=(15, 7), sharex=True)
for i, name in enumerate(DATASETS):
    c = curve.loc[name]
    lv = c.index.values

    # Top row: DIR of the predictions vs. repair level
    ax = axes[0, i]
    ax.errorbar(lv, c[("DIR", "mean")], yerr=c[("DIR", "std")], marker="o", capsize=3, label="DIR (predictions)")
    ax.axhline(before_df.loc[name, "DIR"], color="grey", ls="-.", lw=1, label="DIR (original labels)")
    ax.axhline(1.0, color="green", ls="--", lw=1, label="parity")
    ax.axhline(0.8, color="red", ls=":", lw=1, label="80% rule")
    ax.set_title(name)
    ax.set_ylabel("DIR")

    # Bottom row: predictive performance vs. repair level
    ax = axes[1, i]
    for m in ["Accuracy", "Balanced_Acc"]:
        ax.errorbar(lv, c[(m, "mean")], yerr=c[(m, "std")], marker="o", capsize=3, label=m)
    ax.set_xlabel("repair level λ")
    ax.set_ylabel("score")

axes[0, 0].legend(fontsize=8)
axes[1, 0].legend(fontsize=8)
plt.tight_layout()
plt.savefig("dir_accuracy_vs_repair_level.png", dpi=200)
plt.show()
```
